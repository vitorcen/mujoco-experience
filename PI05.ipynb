## 4. Agent 模式（推荐）

**单行命令（拷贝到终端）：**

GLM 模型：
```bash
ANTHROPIC_BASE_URL=https://open.bigmodel.cn/api/anthropic ANTHROPIC_API_KEY=${ZHIPU_API_KEY} ANTHROPIC_MODEL=glm-4.7 PI05_MODEL_PATH=lerobot/pi05_libero_base python -u pi05_minimax_vla/agent_mode.py
```

MiniMax 模型：
```bash
ANTHROPIC_BASE_URL=https://api.minimaxi.com/anthropic ANTHROPIC_API_KEY=${MINIMAX_API_KEY} ANTHROPIC_MODEL=MiniMax-M2.1 PI05_MODEL_PATH=lerobot/pi05_libero_base python -u pi05_minimax_vla/agent_mode.py
```

**或点击运行下方 cell 在新终端启动**

````python
import subprocess
import os

def run_agent_mode():
    """在新终端窗口中启动 Agent 模式"""
    display = os.environ.get('DISPLAY', ':2')
    model_path = os.environ.get('PI05_MODEL_PATH', 'lerobot/pi05_libero_base')
    # base_url = "https://api.minimaxi.com/anthropic"
    # api_key = os.environ.get('MINIMAX_API_KEY', 'your-api-key')
    # model = "MiniMax-M2.1"
    base_url = "https://open.bigmodel.cn/api/anthropic"
    api_key = os.environ.get('ZHIPU_API_KEY', 'your-api-key')
    model = "glm-4.7"
    
    cmd = f"""gnome-terminal -- bash -c '
export DISPLAY={display}
export PI05_MODEL_PATH={model_path}
export ANTHROPIC_BASE_URL={base_url}
export ANTHROPIC_API_KEY={api_key}
export ANTHROPIC_MODEL={model}

cd {os.getcwd()}
echo "=== Pi05 Agent Mode ==="
echo "环境变量已设置，正在启动..."
echo ""

set +e
python -u pi05_minimax_vla/agent_mode.py
EXIT_CODE=$?

echo ""
if [ $EXIT_CODE -ne 0 ]; then
    echo "❌ 错误：程序异常退出 (退出码: $EXIT_CODE)"
    echo "   请检查上方的错误信息"
else
    echo "✓ 程序正常退出"
fi
echo ""
echo "按 Enter 键关闭窗口..."
read
'"""
    
    subprocess.Popen(cmd, shell=True)
    print("✓ 已在新终端中启动 Agent 模式")
    print(f"  Base URL: {base_url}")
    print("  请切换到新打开的终端窗口进行交互")

run_agent_mode()
````
